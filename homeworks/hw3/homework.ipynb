## Часть 1. KNN Classifier с нуля

KNN - алгоритм instance-based learning. "Обучение" сводится к запоминанию всего train set. Основная работа - в `predict`:

1. Для каждого объекта X_test вычислить расстояния до всех объектов X_train
2. Найти K ближайших
3. Взять мажоритарный класс среди K соседей

**Подсказки:**
- Евклидово расстояние: $d(x, y) = \sqrt{\sum_i (x_i - y_i)^2}$
- Для эффективного поиска K ближайших используйте `np.argpartition` (O(n)) вместо `np.argsort` (O(n log n))
- Мажоритарное голосование: `np.bincount` + `np.argmax`

```python
class KNNClassifier:
    def __init__(self, n_neighbors=5):
        self.n_neighbors = n_neighbors

    def fit(self, X, y):
        """
        "Training" is just memorizing the data.
        """
        self.X_train_ = np.asarray(X)
        self.y_train_ = np.asarray(y)
        self.classes_ = np.unique(self.y_train_)
        return self

    def predict(self, X):
        """
        For each sample in X:
          1. Compute Euclidean distances to all training samples.
          2. Find indices of n_neighbors smallest distances.
          3. Return the most frequent class among those neighbors.
        """
        X = np.asarray(X)

        dists = np.sqrt(((X[:, None, :] - self.X_train_[None, :, :]) ** 2).sum(axis=2))
        neighbor_idx = np.argpartition(dists, self.n_neighbors - 1, axis=1)[:, :self.n_neighbors]
        neighbor_labels = self.y_train_[neighbor_idx]

        preds = np.array([
            np.bincount(labels).argmax()
            for labels in neighbor_labels
        ])

        return preds
```

## Часть 2. PCA с нуля (eigendecomposition)

PCA находит направления максимальной дисперсии. Алгоритм:

1. **Центрирование**: вычесть среднее по каждому признаку
2. **Ковариационная матрица**: $C = \frac{1}{n-1} X_c^T X_c$
3. **Eigendecomposition**: найти собственные значения и векторы матрицы $C$
4. **Сортировка**: отсортировать по убыванию собственных значений
5. **Выбор**: взять первые `n_components` собственных векторов

**Подсказки:**
- `np.linalg.eigh(C)` для симметричной матрицы (ковариация всегда симметрична) - возвращает собственные значения **в возрастающем порядке**
- Чтобы развернуть: `[::-1]` для значений и соответствующая перестановка столбцов для векторов
- `components_` - матрица (n_components, n_features), где каждая строка - собственный вектор
- `transform(X)`: $X_{new} = (X - \bar{X}) \cdot W^T$, где $W$ = `components_`
- `inverse_transform(X_new)`: $X \approx X_{new} \cdot W + \bar{X}$

```python
class PCA:
    def __init__(self, n_components):
        self.n_components = n_components

    def fit(self, X):
        X = np.asarray(X)
        n_samples = X.shape[0]

        # 1. mean
        self.mean_ = X.mean(axis=0)

        # 2. center
        X_centered = X - self.mean_

        # 3. covariance matrix
        C = (X_centered.T @ X_centered) / (n_samples - 1)

        # 4. eigendecomposition
        eigenvalues, eigenvectors = np.linalg.eigh(C)

        # sort in descending order
        order = np.argsort(eigenvalues)[::-1]
        eigenvalues = eigenvalues[order]
        eigenvectors = eigenvectors[:, order]

        # 5. select top components
        self.explained_variance_ = eigenvalues[:self.n_components]
        self.components_ = eigenvectors[:, :self.n_components].T
        total_variance = eigenvalues.sum()
        self.explained_variance_ratio_ = self.explained_variance_ / total_variance

        return self

    def transform(self, X):
        X = np.asarray(X)
        return (X - self.mean_) @ self.components_.T

    def inverse_transform(self, X_new):
        X_new = np.asarray(X_new)
        return X_new @ self.components_ + self.mean_

    def fit_transform(self, X):
        return self.fit(X).transform(X)
```

### Проверка PCA

```python
if __name__ != "homework":
    # Визуализация PCA на Iris
    iris = load_iris()
    X_iris = iris.data
    y_iris = iris.target

    our_pca = PCA(n_components=2).fit(X_iris)
    X_our = our_pca.transform(X_iris)

    sk_pca = SkPCA(n_components=2).fit(X_iris)
    X_sk = sk_pca.transform(X_iris)

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    for cls in np.unique(y_iris):
        mask = y_iris == cls
        axes[0].scatter(X_our[mask, 0], X_our[mask, 1], label=iris.target_names[cls], s=30)
        axes[1].scatter(X_sk[mask, 0], X_sk[mask, 1], label=iris.target_names[cls], s=30)

    axes[0].set_title(f'Our PCA (explained variance: {our_pca.explained_variance_ratio_.sum():.2%})')
    axes[1].set_title(f'sklearn PCA (explained variance: {sk_pca.explained_variance_ratio_.sum():.2%})')
    for ax in axes:
        ax.set_xlabel('PC1')
        ax.set_ylabel('PC2')
        ax.legend()
        ax.grid(alpha=0.2)
    plt.suptitle('Iris: 4D -> 2D через PCA', fontsize=14)
    plt.tight_layout()
    plt.show()

    # Примечание: наши компоненты могут отличаться знаком от sklearn - это норма.
```

```python
if __name__ != "homework":
    # Scree plot: доля дисперсии каждой компоненты
    digits = load_digits()
    pca_full = PCA(n_components=digits.data.shape[1]).fit(digits.data)

    plt.figure(figsize=(12, 5))
    plt.bar(range(1, len(pca_full.explained_variance_ratio_) + 1),
            pca_full.explained_variance_ratio_, alpha=0.7, label='Individual')
    plt.plot(range(1, len(pca_full.explained_variance_ratio_) + 1),
             np.cumsum(pca_full.explained_variance_ratio_), 'ro-', label='Cumulative')
    plt.axhline(y=0.95, color='g', linestyle='--', alpha=0.5, label='95% threshold')
    plt.xlabel('Component')
    plt.ylabel('Explained variance ratio')
    plt.title('Scree plot (digits, 64 features)')
    plt.legend()
    plt.grid(True, axis='y')
    plt.show()

    n_for_95 = np.argmax(np.cumsum(pca_full.explained_variance_ratio_) >= 0.95) + 1
    print(f'Для 95% дисперсии нужно {n_for_95} компонент (из 64)')
```

## Часть 3. Эксперимент: KNN + PCA

Исследуем, как точность KNN зависит от размерности признакового пространства. KNN страдает от curse of dimensionality: в высоких размерностях расстояния "сплющиваются" и соседи становятся неинформативными.

Гипотеза: PCA-проекция может помочь (убирая шумные измерения), но только до определенного момента - слишком сильное сжатие удаляет полезную информацию.

```python
def experiment_knn_pca(X, y, n_components_list, k=5, test_size=0.3, random_state=42):
    """
    Run KNN after PCA dimensionality reduction for various n_components.
    """
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state,
    )

    accuracies = []

    for n_components in n_components_list:
        pca = PCA(n_components=n_components)
        X_train_pca = pca.fit_transform(X_train)
        X_test_pca = pca.transform(X_test)

        knn = KNNClassifier(n_neighbors=k)
        knn.fit(X_train_pca, y_train)
        y_pred = knn.predict(X_test_pca)

        acc = accuracy_score(y_test, y_pred)
        accuracies.append(acc)

    return {
        'n_components': list(n_components_list),
        'accuracies': accuracies,
    }
```

### Проверка эксперимента

```python
if __name__ != "homework":
    # Digits: 64 признака, 10 классов
    digits = load_digits()
    n_components_list = [1, 2, 5, 10, 20, 30, 40, 50, 64]
    results = experiment_knn_pca(digits.data, digits.target, n_components_list, k=5)

    # Сравним с baseline: KNN на исходных данных (= n_components = 64 в нашем случае)
    plt.figure(figsize=(12, 6))
    plt.plot(results['n_components'], results['accuracies'], 'o-', linewidth=2, markersize=10, label='KNN + PCA')
    
    # Baseline: прямое применение KNN
    X_tr, X_te, y_tr, y_te = train_test_split(digits.data, digits.target, test_size=0.3, random_state=42)
    baseline_acc = accuracy_score(y_te, KNNClassifier(n_neighbors=5).fit(X_tr, y_tr).predict(X_te))
    plt.axhline(y=baseline_acc, color='r', linestyle='--', label=f'Baseline (no PCA): {baseline_acc:.3f}')

    plt.xlabel('n_components')
    plt.ylabel('Test accuracy')
    plt.title('KNN accuracy vs PCA dimensionality (digits, K=5)')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Лучший результат
    best_idx = int(np.argmax(results['accuracies']))
    print(f"Best: n_components={results['n_components'][best_idx]}, "
          f"accuracy={results['accuracies'][best_idx]:.4f}")
    print(f"Baseline (no PCA): {baseline_acc:.4f}")
```
